```python
import cv2, json, joblib, torch, timm, gc, random
import numpy as np
from pathlib import Path
from PIL import Image
from torchvision import transforms
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

ROOT = Path(r"/Users/jathusan/Documents/trashnet.ia")
DATA_DIR = ROOT / "data_2"
MODEL_PATH = ROOT / "models" / "deep" / "vit_pca256_rf.joblib"
PCA_PATH = ROOT / "models" / "deep" / "pca_transformer.joblib"  # You'll need the PCA transformer
LABELS_PATH = ROOT / "labels.json"
OUT_DIR = ROOT / "notebook" / "outputs_detection"
OUT_DIR.mkdir(parents=True, exist_ok=True)
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
device = "cuda" if torch.cuda.is_available() else "cpu"
```

## 2) Function batch_predict()
- Receives a list of image paths.
- For each image :
    - Loads the image using OpenCV.
    - Segments the object using the segment_foreground function.
    - Returns the ROI (isolated object), bounding box, and mask.
    - Extracts ViT features from the segmented region (vit_feats(roi)).
    - Stores additional metadata such as the bounding box and file path.

```python
# expl
def batch_predict(paths, batch_size=32):
    feats_list, meta = [], []
    for p in paths:
        img = cv2.imread(str(p)); roi, bbox, _ = segment_foreground(img)
        feats_list.append(vit_feats(roi))  # 1 x D
        meta.append((p, bbox))
    X = np.vstack(feats_list)  # N x D
    preds = clf.predict(X)
    return preds, meta
```

## 3) Foreground Segmentation and Bounding Box Extraction
- This code isolates the main object in the image before classification.
- It makes the pipeline more robust by trying a simple fast method first, then a more advanced method only if necessary.
- The segmentation helps reduce background noise and improves feature extraction.

```python

def largest_bbox(mask: np.ndarray):
    # Finds all external contours in the binary mask and returns the bounding box of the largest one.
    # Used to locate the main foreground object.
    cnts, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    if not cnts:
        return None
    c = max(cnts, key=cv2.contourArea)
    x,y,w,h = cv2.boundingRect(c)
    return x,y,w,h
    
def pad_bbox(x,y,w,h, W,H, ratio=0.12):
    # Expands the bounding box by a fixed ratio to avoid cropping the object too tightly.
    # Ensures the padded box stays inside the image boundaries.
    px = int(w*ratio); py = int(h*ratio)
    x0 = max(0, x - px); y0 = max(0, y - py)
    x1 = min(W, x + w + px); y1 = min(H, y + h + py)
    return x0, y0, x1-x0, y1-y0
    
def segment_foreground(img_bgr: np.ndarray):
    #Performs automatic foreground segmentation in two steps:
        #Primary method: thresholding + morphological filtering to isolate the object.
        #Fallback: if the first method fails, it uses GrabCut to extract the foreground.
                #Returns the segmented region of interest (ROI), the bounding box, and the binary mask.
    H, W = img_bgr.shape[:2]
    hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
    v = hsv[:,:,2]
    _, th = cv2.threshold(v, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    th = cv2.morphologyEx(th, cv2.MORPH_OPEN, np.ones((5,5), np.uint8), iterations=1)
    th = cv2.morphologyEx(th, cv2.MORPH_CLOSE, np.ones((9,9), np.uint8), iterations=1)
    bb = largest_bbox(th)
    if bb is not None:
        x,y,w,h = bb
        if w*h > 0.02*W*H:  # évite poussières
            x,y,w,h = pad_bbox(x,y,w,h,W,H,ratio=0.15)
            roi = img_bgr[y:y+h, x:x+w].copy()
            return roi, (x,y,w,h), th
    rect = (int(0.08*W), int(0.08*H), int(0.84*W), int(0.84*H))
    mask = np.zeros((H,W), np.uint8)
    bgdModel = np.zeros((1,65), np.float64)
    fgdModel = np.zeros((1,65), np.float64)
    cv2.grabCut(img_bgr, mask, rect, bgdModel, fgdModel, 5, cv2.GC_INIT_WITH_RECT)
    mask2 = np.where((mask==2) | (mask==0), 0, 255).astype('uint8')
    bb = largest_bbox(mask2)
    if bb is None:
    
        return img_bgr.copy(), (0,0,W,H), mask2
    x,y,w,h = pad_bbox(*bb, W,H, ratio=0.10)
    roi = img_bgr[y:y+h, x:x+w].copy()
    return roi, (x,y,w,h), mask2
```

## 4) Label Loading, Feature Extraction, and Image Classification Pipeline
This code loads the class labels, initializes the trained classifier, and sets up a **pretrained Vision Transformer (ViT)** to extract image features. It defines a function that segments the object in an image, **extracts ViT embeddings** from the region of interest, and **predicts the class** using the loaded classifier. The code then **tests the pipeline** on a **few sample images** and performs a **simple validation** by checking predictions on a small subset of the dataset. Finally, it saves visualization images with bounding boxes and predicted labels.


```python
VERBOSE = False

def log(*args, **kwargs):
    if VERBOSE:
        print(*args, **kwargs)

# First, let's properly load and inspect the labels
try:
    with open(LABELS_PATH, "r") as f:
        labels_data = json.load(f)

    # log(f"Labels structure: {type(labels_data)}")
    # log(f"Labels content: {labels_data}")
    
    if isinstance(labels_data, list):
        class_names = labels_data
        idx_to_class = {i: name for i, name in enumerate(class_names)}

    elif isinstance(labels_data, dict):
        # If it's already a mapping of idx to class names
        if all(isinstance(k, int) or str(k).isdigit() for k in labels_data.keys()):
            idx_to_class = labels_data
            class_names = [idx_to_class[i] for i in sorted(idx_to_class.keys())]
        else:
            # If it's class_to_idx mapping
            class_to_idx = labels_data
            idx_to_class = {v: k for k, v in class_to_idx.items()}
            class_names = list(class_to_idx.keys())
    else:
        raise ValueError(f"Unexpected labels format: {type(labels_data)}")
    
    class_to_idx = {name: idx for idx, name in idx_to_class.items()}

    # log(f"Class names: {class_names}")
    # log(f"Class to index: {class_to_idx}")
    # log(f"Index to class: {idx_to_class}")

except FileNotFoundError:
    print(f"Error: Labels file not found at {LABELS_PATH}")
    exit(1)
except Exception as e:
    print(f"Error loading labels: {e}")
    exit(1)

# Load classifier
try:
    clf = joblib.load(MODEL_PATH)
    # log("Loaded classifier")
except FileNotFoundError as e:
    print(f"Error loading model: {e}")
    exit(1)

# Load ViT model
try:
    vit_model = timm.create_model('vit_base_patch16_224', pretrained=True)
    vit_model.head = torch.nn.Identity()
    vit_model.eval()
    vit_model.to(device)
except Exception as e:
    print(f"Error loading ViT model: {e}")
    exit(1)

# Define transform
vit_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=(0.5, 0.5, 0.5), std=(0.5, 0.5, 0.5)),
])

def vit_feats(img_bgr):
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    pil_img = Image.fromarray(img_rgb)
    x = vit_transform(pil_img).unsqueeze(0).to(device)
    with torch.no_grad():
        feats = vit_model(x).cpu().numpy()
    return feats

def classify_image(path: Path, save_vis=True):
    img = cv2.imread(str(path))
    if img is None:
        raise ValueError(f"Image illisible: {path}")
    roi, (x,y,w,h), mask = segment_foreground(img)
    
    # Extract features
    feats = vit_feats(roi)
    
    n_features = feats.shape[1]
    n_expected = clf.n_features_in_ if hasattr(clf, 'n_features_in_') else None
    
    # log(f"Features extracted: {n_features}D, Classifier expects: {n_expected}D")
    
    # If dimensions don't match, we might need PCA
    if n_expected and n_features != n_expected:
        # log(f"Warning: Feature dimension mismatch! Using first {n_expected} features")
        feats = feats[:, :n_expected]  # Simple workaround - use first n features
    
    pred = int(clf.predict(feats)[0])
    label = idx_to_class[pred]
    
    vis = img.copy()
    cv2.rectangle(vis, (x,y), (x+w, y+h), (0,255,0), 3)
    cv2.putText(vis, label, (x+10, y-10 if y>25 else y+h+30),
                cv2.FONT_HERSHEY_SIMPLEX, 1.0, (0,255,0), 3)
    if save_vis:
        out_path = OUT_DIR / f"{path.parent.name}__{path.stem}__{label}.jpg"
        cv2.imwrite(str(out_path), vis)
    return label, vis, roi, mask

# Test with a few samples first
log("\nTesting with sample images...")

samples = []
for c in class_names[:2]:
    class_dir = DATA_DIR / c
    if class_dir.exists():
        imgs = sorted(class_dir.glob("*.*"))[:1]
        samples += imgs
        log(f"Found {len(imgs)} images for class {c}")
    else:
        log(f"Warning: Class directory {class_dir} not found")

if not samples:
    log("No sample images found. Checking data directory structure...")
    if DATA_DIR.exists():
        log("Contents of data directory:")
        for item in DATA_DIR.iterdir():
            log(f"  - {item.name}")

for p in samples:
    try:
        lbl, vis, roi, mask = classify_image(p, save_vis=True)
        log(f"Processed {p.name} -> {lbl}")
    except Exception as e:
        log(f"Error processing {p}: {e}")

if samples:
    log("\nRunning validation...")
    val_paths, y_true, y_pred = [], [], []
    for c in class_names:
        class_dir = DATA_DIR / c
        if class_dir.exists():
            all_imgs = sorted(class_dir.glob("*.*"))
            if all_imgs:
                k = max(1, int(0.2 * len(all_imgs)))
                for p in all_imgs[-k:]:
                    val_paths.append(p)
                    y_true.append(class_to_idx[c])

    if val_paths:
        for p in val_paths:
            try:
                pred_label, _, _, _ = classify_image(p, save_vis=False)
                y_pred.append(class_to_idx[pred_label])
            except Exception as e:
                log(f"Error in validation for {p}: {e}")

        y_true = np.array(y_true)
        y_pred = np.array(y_pred)

        acc = (y_true == y_pred).mean()
        log(f"Validation (split simple) — Accuracy: {acc:.4f}")
        
        from sklearn.metrics import classification_report
        log(classification_report(y_true, y_pred, target_names=class_names, digits=3))
    else:
        print(f"[VALIDATION ERROR] {p.name} -> {e}")

log(f"Visualisations sauvegardées dans: {OUT_DIR}")
```
